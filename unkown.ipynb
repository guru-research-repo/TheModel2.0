```python
import os
import shutil
from pathlib import Path

# Paths - Adjust the base_src if needed
base_src = r"processed_data/salience128-48lp-mag/updated_faces/64_identities"
base_dst = r"familiar_faces_nikita/40_identities"

# Configuration
identities_to_copy = 40
base_images_per_identity = 1
splits = ['train', 'valid', 'test']

def get_base_filename(filename):
    """Extracts '0' from '0_proc1.png' or '7995' from '7995_proc2.png'"""
    return filename.split('_proc')[0]

def create_familiar_dataset():
    for split in splits:
        split_src = os.path.join(base_src, split)
        if not os.path.exists(split_src):
            continue
            
        # Get list of identity folders (Adele, etc.)
        all_identities = sorted([d for d in os.listdir(split_src) if os.path.isdir(os.path.join(split_src, d))])
        # Select the first 10 identities
        selected_identities = all_identities[:identities_to_copy]

        for identity in selected_identities:
            ident_src_path = os.path.join(split_src, identity)
            print("ident_src_path",ident_src_path)
            ident_dst_path = os.path.join(base_dst, split, identity)
            print("ident_dst_path",ident_dst_path)
            os.makedirs(ident_dst_path, exist_ok=True)

            # Get all files and group them by their base ID
            all_files = os.listdir(ident_src_path)
            # A set of unique base image IDs (e.g., {'0', '1', '7995'})
            unique_bases = sorted(list(set(get_base_filename(f) for f in all_files if '_proc' in f)))
            
            # Select first 10 base images
            selected_bases = unique_bases[:base_images_per_identity]
            
            print(f"Copying {identity} in {split}: {len(selected_bases)} base images (and their fixations)...")

            # Copy all files belonging to those selected bases
            for file in all_files:
                if get_base_filename(file) in selected_bases:
                    src_file = os.path.join(ident_src_path, file)
                    dst_file = os.path.join(ident_dst_path, file)
                    shutil.copy2(src_file, dst_file)

if __name__ == "__main__":
    create_familiar_dataset()
    print("\nSuccessfully created familiar_faces dataset.")
```

```python
import os
import shutil
from pathlib import Path

# Paths - Adjust if needed
base_src = r"processed_data/salience128-48lp-mag/updated_faces/64_identities"
base_dst = r"familiar_faces_nikita/40_identities"  # ← your existing folder

# NEW: identities to ADD (does not affect existing ones)
identities_to_add = ["JeffBezos"]

# Configuration
base_images_per_identity = 1
splits = ['train', 'valid', 'test']

def get_base_filename(filename):
    """Extracts '0' from '0_proc1.png' or '7995' from '7995_proc2.png'"""
    return filename.split('_proc')[0]

def add_selected_identities():
    for split in splits:
        split_src = os.path.join(base_src, split)
        if not os.path.exists(split_src):
            continue

        for identity in identities_to_add:
            ident_src_path = os.path.join(split_src, identity)

            # Skip if identity doesn't exist in source
            if not os.path.exists(ident_src_path):
                print(f"⚠️  Missing in source: {identity} ({split})")
                continue

            ident_dst_path = os.path.join(base_dst, split, identity)
            os.makedirs(ident_dst_path, exist_ok=True)

            # --- SAME LOGIC AS YOUR ORIGINAL CODE ---
            all_files = os.listdir(ident_src_path)

            unique_bases = sorted(
                list(set(get_base_filename(f) for f in all_files if '_proc' in f))
            )

            selected_bases = unique_bases[:base_images_per_identity]

            print(f"Adding {identity} in {split}: "
                  f"{len(selected_bases)} base images (and fixations)...")

            # Copy files (overwrite-safe but not destructive)
            for file in all_files:
                if get_base_filename(file) in selected_bases:
                    src_file = os.path.join(ident_src_path, file)
                    dst_file = os.path.join(ident_dst_path, file)

                    # Only copy if file not already present
                    if not os.path.exists(dst_file):
                        shutil.copy2(src_file, dst_file)

if __name__ == "__main__":
    add_selected_identities()
    print("\nSuccessfully added the two identities.")
```

```python
import os
import shutil
from pathlib import Path

# Base directory
BASE = Path("dogs8_sam_processed_data/salience10-48lp-mag/cnn_dogs")

# Source and destination roots
SRC_ROOT = BASE / "8_identities" / "test"
DST_ROOT = BASE / "4_identities" / "test"

# Folders to copy (source → destination mapping)
FOLDERS = [
    "standard schnauzer",
    "Shih-Tzu",
    "Labrador retriever",
    "basset, basset hound"
]

# Set to True if you want to overwrite existing files
OVERWRITE = False


def copy_folder(src_folder: Path, dst_folder: Path):
    """
    Copies all files from src_folder to dst_folder.
    Creates dst_folder if it doesn't exist.
    Does NOT modify source.
    """

    if not src_folder.exists():
        print(f"[WARNING] Source folder does not exist: {src_folder}")
        return

    # Create destination folder if needed
    dst_folder.mkdir(parents=True, exist_ok=True)

    for item in src_folder.iterdir():
        src_item = item
        dst_item = dst_folder / item.name

        if item.is_file():
            if dst_item.exists():
                if OVERWRITE:
                    shutil.copy2(src_item, dst_item)
                    # print(f"[OVERWRITTEN] {dst_item}")
                # else:
                #     print(f"[SKIPPED] Already exists: {dst_item}")
            else:
                shutil.copy2(src_item, dst_item)
                # print(f"[COPIED] {dst_item}")

        elif item.is_dir():
            # Recursively copy subdirectories
            copy_folder(src_item, dst_item)


def main():
    for folder_name in FOLDERS:
        src_path = SRC_ROOT / folder_name
        dst_path = DST_ROOT / folder_name

        print(f"\nProcessing: {folder_name}")
        copy_folder(src_path, dst_path)

    print("\nDone.")


if __name__ == "__main__":
    main()
```
